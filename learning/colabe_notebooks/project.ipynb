# Random cubes and sides:

```
#V1 num generator:
from random import randint
print()
cubes = int(input('Plese enter any number of cubes: '))
sides = int(input('Plese enter any number of sides: '))
print()
for _ in range(1, cubes+1):
  num = randint(1, sides)
  print(f'[{num}]', end =" ")

#show type:
print(f'\n\n{type(num)}')
```

```
#V2 dict output:
import random 

cubes = int(input('Plese enter any number of cubes: '))
sides = int(input('Plese enter any number of sides: '))
print()
s = {}
for d in range(cubes):
  for i in range(1, sides+1):
    s[i] = []
    s[i].append(i)
  print(random.choice(list(s.values())), end =" ")
  
#show type:
print(f'\n\n{type(s)}')
```

```
#V3 list output:
import random 

cubes = int(input('Plese enter any number of cubes: '))
sides = int(input('Plese enter any number of sides: '))
print()
s = []
for d in range(cubes):
  for i in range(1, sides+1):
    if i not in s:
      s.append(i)
  print(f'[{random.choice(s)}]', end =" ")
  
#show type:
print(f'\n\n{type(s)}')
```

# Cover

```
import random
import numpy as np
import matplotlib.pyplot as plt

# 8 square attractors (x,y):
a1, a2, a3, a4, a5, a6, a7, a8 = (0,0), (3,0), (3,3), (0,3), (1,1), (2,1), (2,2), (1,2)
# square spaces 0+0.125=0.125+0.125+...=1:
square_spaces = [0.125, 0.25, 0.375, 0.5, 0.625, 0.75, 0.875, 1.0]
# square points on (x,y):
q0 = [a1, a2, a3, a4, a5, a6, a7, a8]

# generate random (x, y) for p0:
x_1 = random.uniform(0, 3)
y_1 = random.uniform(0, 3)

# start point in q0:
p0 = [x_1, y_1]

#number of iterations:
for t in range(1000):
  n = random.uniform(0, 1)
  for index, num in enumerate(square_spaces):
    if num-0.125 <= n < num:
      # aa is active attractor:
      aa = q0[index]
      xa, ya = aa[0], aa[1]
      xi = (p0[0] + 2*xa)/3
      yi = (p0[1] + 2*ya)/3
      pi = (round(xi, 3), round(yi, 3))
      p0[0] = xi
      p0[1] = yi
      # print('n =', round(n, 2))
      # print('aa =', aa)
      # print('pi =', pi, '\n')
      plt.scatter(xi, yi)
    elif 0.875 <= n <= 1:
      aa = q0[7]
      xa, ya = aa[0], aa[1]
      xi = (p0[0] + 2*xa)/3
      yi = (p0[1] + 2*ya)/3
      pi = (round(xi, 3), round(yi, 3))
      p0[0] = xi
      p0[1] = yi
      plt.scatter(pi[0], pi[1])
    else:
      continue
plt.show()
```
