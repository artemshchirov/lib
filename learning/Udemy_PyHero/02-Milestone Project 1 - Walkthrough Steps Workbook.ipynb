```python
def player_input():
    
    choice = ''
    while (choice not in ['X', 'O']):

      choice = input('Input X or O: ').upper()
      
      if (choice not in ['X', 'O']):
        print('Just X or O!')
        continue
      else:
        if choice == 'X':
          return 'X'
        else:
          return 'O'
```

```python
def space_check(board, position):
    
    if board[position] == ' ':
      return True
    else:
      return False
```

```python
from random import randint

def choose_first():
    
    if randint(0, 1) == 1:
      print('First X')
      return 'X'
    else:
      print('First O')
      return 'O'
```

```python
def replay():
    
  choice = input('Play again? (Y / N)')
  while (choice not in ['Y', 'N']):
    choice = input('Play again? (Y / N)')

  if (choice == 'Y'):
    return True
  elif (choice == 'N'):
    return False
```
