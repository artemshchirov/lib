```
def xyz_there(str):

  countxyz = 0
  count_xyz = 0
  for index in range(len(str)-2):
    if str[index] == 'x' and str[index+1] == 'y' and str[index+2] == 'z':
      countxyz += 1
    elif str[index] == '.' and str[index+1] == 'x' and str[index+2] == 'y' and str[index+3] == 'z':
      count_xyz += 1

  return countxyz > count_xyz

xyz_there('abc.xyzxyz')
```

```
def count_evens(nums):
  even = 0
  for i in nums:
    if i % 2 == 0:
      even += 1
  return even

count_evens([1, 2, 3, 4])
```

```
def big_diff(nums):
  return max(nums) - min(nums)

big_diff([1,2,3,4,5,-7])
```

```
def centered_average(nums):
  nums = set(nums)
  return (sum(nums) - (max(nums) + min(nums))) // (len(nums)-2)

centered_average([-10, -4, -2, -4, -2, 0]) 
```

```
def has22(nums):

  for index in range(len(nums)-1):
    if nums[index] == 2 and nums[index+1] == 2:
      return True
    else:
      continue
  return False

has22([1, 2, 1, 2, 2])
```

```
def sum67(nums):

  flag = True
  count = 0
  for num in nums:

    if num == 6:
      flag = False
    elif flag:
      count += num
    elif num == 7:
      flag = True
  
  return count

sum67([1, 2, 2, 6, 99, 99, 7])
```

```
def string_bits(str):
  new_str = []
  for i, l in enumerate(str):
    if i % 2 == 0:
      new_str.append(l)
    else:
      continue
  return new_str
           
string_bits('Heeololeo')
```

```
def string_splosion(str1):
  result = ''
  for letter in range(len(str1)):
    result += str1[:letter+1]
  return result

string_splosion('Code')
```

```
def last2(str1):

  count = 0
  for i in range(len(str1)-3):
    if str1[i] == str1[0] and str1[i+1] == str1[1]:
      count += 1
    else:
      continue
  return count
last2('hixxxhi')
```

# lambda, map, filter 

```
my_nums = [1, 2.5, 3, 4, 5, -6]
def square(num):
  return num**2

# map - function iterate map(func, iterable):
print(list(map(square, my_nums)))

for item in map(square, my_nums):
  print(item)
```

```
my_string = ['Artem', 'Nanna', 'WHOAREYOU?', '0010110101']
def splicer(mystring):
  if len(mystring) % 2 == 0:
    return 'EVEN'
  else:
    return mystring[0]

print(splicer(my_string))

list(map(splicer, my_string))
```

```
my_nums = [1, 2, 3, 4, 5, -6]
def check_even(num):
  return num % 2 == 0

# 'filter' use func to iterable and return just if TRUE:
for item in filter(check_even, my_nums):
  print(item)

# or print from filter to list:
list(filter(check_even, my_nums))
```
