```
# ---------------- Ситуационные функции и методы:
# Функция вввода значения
def getInput(digit, message):

  ret = ''
  while (ret == '' or not ret in digit):
    ret = input(message)
    
  return ret
```

```
# Функция ввода целого числа
def getIntInput(min, max, message):

  ret = -1
  while (ret < min or ret > max):
    str1 = input(message)
    if str1.isdigit():
      ret = int(str1)
    else:
      print(colored('    Введите целое число!', 'cyan'))

  return ret
```

```
import time
import random

def getDice():
# --------------- Бросок, получение значений и анимация двух кубов

  count = random.randint(3, 8)           # Сколько раз кубы будут показывать значения 
  sleep = 0                              # Время пауз между показом значений кубов
  col = 0                                # Номер цвета в colors[col]
  colors = ['cyan', 'red', 'magenta',    # Список цветов для строк рулетки
          'blue', 'green', 'yellow']

# Цикл показа знчаений кубов пока count > 0
  while (count > 0):
    
    x = random.randint(1, 6)    # Генерация числа куба в данную итерацию
    y = random.randint(1, 6)

    print(colored(f'{" " * 10}----- -----', colors[col]))
    print(colored(f'{" " * 10}| {x} | | {y} |', colors[col]))
    print(colored(f'{" " * 10}----- -----', colors[col]))

# Изменение цвета в colors[col]
    col += 1
    if (col == 6):
      col = 0

# Пауза между показом значений кубов
    time.sleep(sleep)
# Увеличение паузы между показом значений кубов
    sleep += 1 / count

# Уменьшение количества итераций в цикле
    count -= 1

  return x + y

# getDice()
```

```
def getOHBRes(stavka):

  res = stavka
  
# Функция определения значений 5-ти слотов машины и показ анимации 
  d1 = 0                                  # 5 окон слот машины, цифра - значение слота номер d
  d2 = 0
  d3 = 0
  d4 = 0
  d5 = 0

  getD1 = True                            # Логические переключатели для d1-d5
  getD2 = True                            # Пока True: d1-d5 меняют значения (d1 + 1, d2 - 1, ..., d5 +1)
  getD3 = True
  getD4 = True
  getD5 = True

  col = 0                                 # Номер цвета в colors[col]
  colors = ['cyan', 'red', 'magenta',     # Список цветов для смены цвета строк каждую итерацию в цикле
            'blue', 'green', 'yellow']

  while (getD1 == True
         or getD2 == True
         or getD3 == True
         or getD4 == True
         or getD5 == True):
    
# Смена цифр в слотах (d1 + 1, d2 - 1, ..., d5 +1)
    if (getD1):
      d1 += 1
    if (getD2):
      d2 -= 1
    if (getD3):
      d3 += 1
    if (getD4):
      d4 -= 1
    if (getD5):
      d5 += 1

# Ограничение значений в слотах от 0 до 9
    if (d1 > 9):
      d1 = 0
    if (d2 < 0):
      d2 = 9
    if (d3 > 9):
      d3 = 0
    if (d4 < 0):
      d4 = 9
    if (d5 > 9):
      d5 = 0

# С шансом 1/y+1 в каждой итерации остановить смену значения слотов 
    if (random.randint(0, 10) == 1):    # (0, 10) == (x, y)
      getD1 = False
    if (random.randint(0, 10) == 1):
      getD2 = False
    if (random.randint(0, 10) == 1):
      getD3 = False
    if (random.randint(0, 10) == 1):
      getD4 = False
    if (random.randint(0, 10) == 1):
      getD5 = False

# Пауза между итерациями для имитации слот- машины
    time.sleep(0.1)

# Строка со знаком % и слотами, оформление
    print(colored(f'    {"% " * 11}', colors[col]))
    print(colored(f'    | {d1} | {d2} | {d3} | {d4} | {d5} |', colors[col]))
    print(colored(f'    {"- " * 11}', colors[col]))

# Изменение цвета в colors[col]
    col += 1
    if (col == 6):
      col = 0

  maxCount = getMaxCount(d1, d1, d2, d3, d4, d5)

  if (maxCount < getMaxCount(d2, d1, d2, d3, d4, d5)):
      maxCount = getMaxCount(d2, d1, d2, d3, d4, d5)
  if (maxCount < getMaxCount(d3, d1, d2, d3, d4, d5)):
      maxCount = getMaxCount(d3, d1, d2, d3, d4, d5)
  if (maxCount < getMaxCount(d4, d1, d2, d3, d4, d5)):
      maxCount = getMaxCount(d4, d1, d2, d3, d4, d5)
  if (maxCount < getMaxCount(d5, d1, d2, d3, d4, d5)):
      maxCount = getMaxCount(d5, d1, d2, d3, d4, d5)

  if (maxCount == 2):
    print(colored(f' Совпадение двух чисел! Твой выигрыш в размере ставки: {res}','yellow'))
  elif (maxCount == 3):
    print(colored(f' Совпадение трёх чисел! Твой выигрыш 2:1: {res}','yellow'))
    res = stavka * 2
  elif (maxCount == 4):
    print(colored(f' Совпадение ЧЕТЫРЕХ чисел! Твой выигрыш 5:1: {res}','yellow'))
    res = stavka * 5
  elif (maxCount == 5):
    print(colored(f' Совпадение ВСЕХ чисел! Твой выигрыш 10:1: {res}','yellow'))
    res = stavka * 10
  else:
    loose(res)
    res = 0
    
  input(' Нажми Enter для продолжения... ')

  return res

# getOHBRes()
```

```
def getMaxCount(digit, v1, v2, v3, v4, v5):

  ret = 0
  
  if (digit == v1):
    ret += 1
  if (digit == v2):
    ret += 1
  if (digit == v3):
    ret += 1
  if (digit == v4):
    ret += 1
  if (digit == v5):
    ret += 1

  return ret
```
