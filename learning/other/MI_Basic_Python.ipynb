#Classes
### Что такое class?
 - чертеж объекта, с которым потом можно будет работать
 - Внутрь можно добавить func, values. Все это будет одним классом
 - На его основании потом можно создать obj
 - method в class = func
 - obj по инструкции = экземпляр класса 

```
# Create an class:
# Принято называть classes с большой буквы:
class GameCharacter():

# Simple model of player character:
  # Attributes:
    # name
    # heaslth
    # x_pos

# Static variable:
  GameCharacter.speed = 1.0

# method '__init__' - автоматически исполняется
# при создании каждого нового экземпляра класса. 
# Во время вызова исполняет всё, что в нём было записано  ранее:
 # method 'self' - ссылка на экземпляр класса (здесь - PlayerCharacter:)
 # описанного всем, что после его ':'.
 # Это ссылка, что работа будет с ним.
  def __init__(self, name, health, x_pos):
    
# Initial attributes: 
    self.name = name
    self.health = health
    self.x_pos = x_pos
    print('GameCharacter created')

# PlayerCharacter will sit down:
  def sit(self):
 # Start method 'sit':
    print(self.name.title() + ' sit down')

# PlayerCharacter will jump:
  def jump(self):
 # Start method 'jump':
    print(self.name.title() + ' allez hop!')

# Player character will move:
  def move(self, by_amount):
    self.x_pos += by_amount

# Player character will take damage:
  def take_damage(self, amount):
    self.health -= amount
    if self.health < 0:
      self.health = 0

  def check_is_dead(self):
    return self.health <= 0

  def change_speed(to_speed):
    GameCharacter.speed = to_speed

# Создание экземпляра класса:
 # 1st character:
fir_char = GameCharacter('Neuron', 100, 0)
 # 2st character:
Atom = GameCharacter('Atom', 75, 4)

print()
print('Neuron:', fir_char.health, fir_char.x_pos)
print('Atom:', Atom.health, Atom.x_pos)

# Use, return methods:
print()
fir_char.sit()
Atom.jump()
```

```
# Create an subclass of class:
class PlayerCharacter(GameCharacter):

  def __init__(self, name, health, x_pos, num_lives):
    super().__init__(name, health, x_pos)
 #  super().move(x) <-- work`s too
    self.max_health = health
    self.num_lives = num_lives

  def take_damage(self, amount):
    self.health -= amount
    if self.health <= 0:
      self.num_lives -= 1
      self.health = self.max_health

  def check_is_dead(self):
    return self.health <= 0 and num_lives <= 0

pc = PlayerCharacter('Quark', 100, 4, 3)
gc = GameCharacter('Proto', 50, 0)

print(pc.x_pos)
pc.move(2)
print(pc.x_pos)
print()
print(gc.x_pos)
gc.move(-2)
print(gc.x_pos)

print()
print(pc.health)
pc.take_damage(50)
print(pc.health)
print(pc.check_is_dead())
print(pc.num_lives)
print()
print(gc.health)
gc.take_damage(200)
print(gc.health)
print(gc.check_is_dead())
```

```
class PlayerInventory(PlayerCharacter):

  def __init__(self, name, health, x_pos, num_lives, inventory = 5):
    super().__init__(name, health, x_pos, num_lives)
    self.inventory = inventory
    self.max_inventory = 10

  def take(self):
    if self.inventory < self.max_inventory:
      self.inventory += 1

  def give(self):
    if self.inventory > 0:
      self.inventory -= 1


pi = PlayerInventory('Quark', 100, 5, 3, 4)
ni = PlayerInventory('NPC', 50, 2, 1)

print()
print(pi.inventory)
pi.give()
print(pi.inventory)
pi.give()
print(pi.inventory)
pi.give()
print(pi.inventory)
pi.give()
print(pi.inventory)
pi.give()
print(pi.inventory)
print()
print(ni.inventory)
ni.take()
print(ni.inventory)
ni.take()
print(ni.inventory)
ni.take()
print(ni.inventory)
ni.take()
print(ni.inventory)
ni.take()
print(ni.inventory)
ni.take()
print(ni.inventory)
```
