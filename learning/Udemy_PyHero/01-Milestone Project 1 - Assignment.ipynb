```python
def side_choice():

  print('-'*50 + '\nPlayer 1: Do you want to be X or O?')

  player_1 = 'unknown'

  while player_1 not in ['X', 'O']:
    
    player_1 = input()

    if player_1 not in ['X', 'O']:
      print('Please, enter X or O: ')

    if player_1 == 'X':
      print('-'*50 + '\nPlayer 1 will go first')
      return 'O'
    elif player_1 == 'O':
      print('-'*50 + '\nPlayer 2 will go first')
      return 'X'
```

```python
def ready():

  print('-'*50, '\nAre you ready to play the GAME? Enter Yes or No')
  yesOrNo = 'Yes or No?'

  while (yesOrNo not in ['Yes', 'No']):

    yesOrNo = input()

    if (yesOrNo == '0'):
      return True
    elif (yesOrNo not in ['Yes', 'No']):
      print('Please, enter Yes or No: ')

    if (yesOrNo == 'Yes'):
      return False
    elif (yesOrNo == 'No'):
      return True
```

```python
def display_board(boardInputs):

  boardDisplay ="""
       |     |
    {}  |  {}  |  {}
  _____|_____|_____
       |     | 
    {}  |  {}  |  {}
  _____|_____|_____
       |     |
    {}  |  {}  |  {}
       |     |
  """
  
  print(boardDisplay.format(*boardInputs))
```

```python
def player_choice(symbol):

  choice = 'noDigit'               # Coordinates for the symbol in board output
  acceptableRange = range(1,10)    # Check input if number 1 - 9
  withinRange = False              # If True - input in acceptable range

# --------------- TWO CONDITIONS TO CHECK
  while (not choice.isdigit()) or (not withinRange):

    choice = input('Choose your next position (1-9): \n (for exit input 0) ')
    
    if choice == '0':    # input 0 for exit
      return True

# --------------- DIGIT CHECK
    if (not choice.isdigit()):
      print('Please, enter a number (1-9): ')
      continue

# --------------- RANGE CHECK
    if (choice.isdigit()):
      if int(choice) in acceptableRange:
        withinRange = True
      else:
        print('Sorry you are out of acceptable range (1-9)')
  
  choice = int(choice)

  if boardInputs[choice-1] == ' ':    # Check if choosen location are empty
    boardInputs[choice-1] = symbol
  else:
    player_choice(symbol)             # Try once more
  
  if ' ' not in boardInputs:          # Check for tie
    print('Tie!')
    return True                       # win = True

# --------------- CHECK FOR WIN                                                                                      
  if (boardInputs[:3] == ['X', 'X', 'X'] or boardInputs[:3] == ['O', 'O', 'O']            # 1, 2, 3
    or boardInputs[3:6] == ['X', 'X', 'X'] or boardInputs[3:6] == ['O', 'O', 'O']         # 4, 5, 6
    or boardInputs[6:9] == ['X', 'X', 'X'] or boardInputs[6:9] == ['O', 'O', 'O']         # 7, 8, 9
    or boardInputs[:7:3] == ['X', 'X', 'X'] or boardInputs[:7:3] == ['O', 'O', 'O']       # 1, 4, 7
    or boardInputs[1:8:3] == ['X', 'X', 'X'] or boardInputs[1:8:3] == ['O', 'O', 'O']     # 2, 5, 8
    or boardInputs[2:9:3] == ['X', 'X', 'X'] or boardInputs[2:9:3] == ['O', 'O', 'O']     # 3, 6, 9
    or boardInputs[:9:4] == ['X', 'X', 'X'] or boardInputs[:9:4] == ['O', 'O', 'O']       # 1, 5, 9
    or boardInputs[2:7:2] == ['X', 'X', 'X'] or boardInputs[2:7:2] == ['O', 'O', 'O']):   # 3, 5, 7
    clear_output()
    display_board(boardInputs)  
    print('Congratulations! {} have won THE GAME!'.format('Player 1' if symbol == 'X' else 'Player 2'))  
    return True    
```
